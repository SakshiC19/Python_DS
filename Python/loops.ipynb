## Loops

```python
i=0
while i<5:
    print(i)
    i+=1
```

```python
i=0
while i<5:
    i+=1
    print(i)
```

```python
#to print even numbers using while loop from 1 to 15

i=2
while i<15:
    print(i)
    i+=2
```

```python
#to print odd numbers using while loop from 1 to 15

i=1
while i<15:
    print(i)
    i+=2
```

### break exists the loop if the condition is true
### continue will skip the condition and continue the further condition

```python
i=0
while i<5:
    i=i+1
    if i==3:
        continue
    print(i)
```

```python
i=0
while i<5:
    i=i+1
    if i==3:
        break
    print(i)
```

## For Loop

```python
num=[23,1,24,12,23,1,14,5,67,65,68,99,62,94,20]
```

```python
for i in num:
    print(i)
```

```python
# try to exact even numbers from this list

for i in num:
    if i%2==0:
        print(i)
```

```python
# try to exact odd numbers from this list

for i in num:
    if i%2 !=0:
        print(i)
```

```python
for i in num:
    if i%2==0:
        print(i,"Is Even")
    else:
        print(i,"Is Odd")
```

```python
#to print 0 to 9 in range()

for i in range(10):                     #i+1 value in range()
    print(i)
```

```python
for i in range(2,11,2):         #(start_with,end_value+1,incre/decre)
    print(i)
```

```python
for i in range(1,11,2):
    print(i)
```
