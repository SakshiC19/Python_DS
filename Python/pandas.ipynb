### 2. Creating DataFrames and Series

##### pd.Series() - create a 1D labeled array 

```python
s=pd.Series([10,20,30])
s
```

```python
s=pd.Series([10,20,30],index=['a','b','c'],name='values')
s
```

##### pd.DataFrame() - from a dict of lists - create a labeled 2D table 

```python
df = pd.DataFrame({
    'name': ['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','Goa']
})
df
```

##### pd.DataFrame() - from a list of dicts

```python
pd.DataFrame([{'a':1,'b':2},{'a':2,'b':4}])
```

##### pd.DataFrame() - from a NumPy array

```python
pd.DataFrame(np.arange(6).reshape(2,3),columns=['x','y','z'])           #reshape in 2 rows and 3 col with col name x y and z and array value startwith 0 to 5
```
