```python
#try to write a code to check word is palindrome or not

word =str(input("Enter a word :"))

# Convert to lowercase for case-insensitive comparison
word=word.lower()   

if word==word[::-1]:
    print("Is a palindrome")
else:
    print("Not a palindrome")
    
```

```python
#Another way to check palindrome or not but o/p in true or false

def palindrome(word):
    word=word.lower()
    return word==word[::-1]
palindrome("mOM")
```

```python
# write a code to check number is even or odd

num=int(input("Enter any number:"))
def even_odd(num):
    if num%2==0:
        print("It is Even")
    else:
        print("It is Odd")
even_odd(num)
        
```

```python
# write a function to check number is positive or negative

def pos_neg(num):
    if num>0:
        print("It is an Positive Number")
    else:
        print("It is an Negative Number")

num=int(input("Enter any integer:"))
pos_neg(num)
```
