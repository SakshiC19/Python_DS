```python
#.prod() - product of all element

arr.prod()
```

### 8. Sorting & Searching

```python
arr=np.array([3,1,4,1,5,9,2])
arr
```

```python
#.sort() - return a sorted copy

np.sort(arr)
```

```python
#.argsort() - indices that would sort the array

arr.argsort()                #gives index value of sorted array
```

```python
#.np.searchsorted()  - index where to insert a value and keep order

np.searchsorted(np.sort(arr),4)      #tells you index value to where you can insert the value 
```

```python
np.where(arr==1)        #wherever your condition is present it will give you a index value of it
```

```python
#np.nonzero(arr) - indices of nonzero elements

np.nonzero(arr)
```

```python
#np.unique() - sorted unique values
np.unique(arr)
```

```python
#np.unique(arr,return_counts=True) - unique values with their counts

np.unique(arr,return_counts=True)
```

```python
#np.extract() - extract elements matching a condition

np.extract(arr>3,arr)       #array greater than 3
```
