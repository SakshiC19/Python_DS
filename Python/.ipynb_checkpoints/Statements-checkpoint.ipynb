```python
#Rating system
#1-poor service , #2-can be better #3- good service #4-very good service #5- excellent service

rate = int(input("Enter Rating: "))

if rate == 1:
    print("Poor Service")
elif rate == 2:
    print("Can Be Better")
elif rate == 3:
    print("Good Service")
elif rate == 4:
    print("Very Good Service")
elif rate == 5:
    print("Excellent Service")
else:
    print("Invalid Rating")
```

```python
#Grading System
#Above 90 -A+ , Above 80 -A , Above 70 -B , Above 60 -C , Above 50 -D , Below 50 - Fail 

grade =int(input("Enter Your Grade :"))

if grade>90 and grade<=100:
           print("A+")
elif grade>80 and grade<=90:
            print("A")
elif grade>70 and grade<=80:
           print("B")
elif grade>60 and grade<=70:
           print("C")
elif grade>=50 and grade<=60:
           print("D")
elif grade<50 :
            print("Fail")
else:
    print("Invalid Grade")
```

```python
# if age<18 - minor ,age between 18 to 60 - Adult ,
#age between 60 to 100 - Senior Citizen  else Invalid

age=int(input("Enter your age"))
if age<18 and age>0:
    print("Minor")
elif age>=18 and age<=60:
    print("Adult")
elif age>60 and age<=100:
    print("Senior Citizen")
else:
    print("Invalid")
```

```python
#Bank loan approval -> Country -> Employement -> Salary -> Credit score(above 30,000)


country=str(input("Enter your Country:"))
emp_status=str(input("Enter your status(Employed/Unemployed):"))
salary=int(input("Enter your salary:"))
credit_score=int(input("Enter your credit score:"))

if country.upper()=="INDIAN" or country.upper()=="INDIA" :
    if emp_status.upper()=="EMPLOYED":
        if salary>=30000:
            if credit_score>=700:
                print("Loan Approved")
            else:
                print("Loan Rejected Low Credit Score")
        else:
            print("Loan Rejected Unsufficient Salary")
    else:
        print("Loan Rejected : Unemployed")
else:
    print("Loan Rejected : Country Not Supported")
```

```python
#College Admission - State- Age-Qualification- Marks

state=str(input("Enter your State:"))
age=int(input("Enter your Age:"))
qualification=str(input("Enter your Qualification(12th Pass/Diploma):"))
marks=int(input("Enter your Marks:"))

if state.lower() == "maharashtra":
    if age >= 18 and age <= 24:
        if qualification == "12th Pass" or qualification =="12th pass":
            if marks >= 60:
                print("Admission Confirmed ")
            else:
                print("Admission Rejected: Marks below cutoff")
        else:
            print("Admission Rejected: Qualification Not Eligible")
    else:
        print("Admission Rejected: Age Not in Eligible Range (18-24)")
else:
    print("Admission Rejected: State Not Supported")
    
```

```python
# Online shopping Example → Login → Product availability → Payment → Order confirmation

username=str(input("Enter Username:"))
password=str(input("Enter Password:"))
product_available=str(input("Is the product in stock? (Yes /No):"))
quantity = int(input("Enter quantity required: "))
stock_quantity = int(input("Enter available stock quantity: "))
payment_status = str(input("Enter payment status (Success/Failed): "))

# Correct credentials 
correct_username ="admin"
correct_password ="1234"


if username == correct_username and password == correct_password:
    if product_available.upper() =="YES":
        if quantity <= stock_quantity:
            if payment_status.upper() =="SUCCESS":
                print("Order Confirmed! Thank you for shopping with us.")
            else:
                print("Order Failed: Payment Not Successful")
        else:
            print("Order Failed: Insufficient Stock")
    else:
        print("Order Failed: Product Not Available")
else:
    print("Login Failed: Invalid Username or Password")
```

```python
#ATM withdrawal Example → Card valid → PIN correct → Balance available → Withdraw

card_status=str(input("Is your card valid(Yes/No):"))
pin=int(input("Enter your PIN:"))
correct_pin=1234
balance=int(input("Enter your account balance:"))
withdraw_amount=int(input("Enter amount to withdraw:"))

if card_status.upper()=="YES":
    if pin==correct_pin:
        if balance>=withdraw_amount:
            balance-=withdraw_amount                     # balance= balance - withdraw_amount
            print("Withdrawal Successful.")
            print("Remaining Balance",balance)
        else:
            print("Transition Failed : Insufficient Balance")
    else:
        print("Transition Failed : Incorrect PIN")
else:
    print("Transition Failed:Invalid Card")
```

```python
#Insurance eligibility → Country → Age → Income → Medical/other criteria	


country =str(input("Enter your country :"))
age=int(input("Enter your age :"))
income=int(input("Enter your Annual Income :"))
medical_condition=str(input("Do you have any medical Condition?(Yes/No):"))

if country.upper()=="INDIA":
    if age>=18 and age<=60:
        if income>=200000:
            if medical_condition.upper()=="NO":
                print("Insurance Approved")
            else:
                print("Insurance Rejected : Due to Medical Condition")
        else:
            print("Insurance Rejected : Income Below Minimum Required")
    else:
        print("Insurance Rejected : Age Not in Eligible Criterior(18-60)")
else:
    print("Insurance Rejected : Country Not Supported")
    
```
